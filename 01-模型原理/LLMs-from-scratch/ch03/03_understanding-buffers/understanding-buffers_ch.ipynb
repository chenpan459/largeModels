# 理解 PyTorch Buffers

PyTorch buffer 是与模块/模型关联的张量属性，类似 parameter，但训练时不更新。

Buffer 在 GPU 计算中特别有用：需与 parameter 一起在设备间（如 CPU→GPU）迁移。Buffer 不需要梯度，但必须在正确设备上以保证计算正确。

第 3 章通过 `self.register_buffer` 使用 buffer，书中仅简要说明。本 notebook 用动手示例详细解释。

## 无 buffer 的示例

以下代码基于第 3 章，已修改为不含 buffer，实现 LLM 使用的因果自注意力：

```python
import torch
import torch.nn as nn

class CausalAttentionWithoutBuffers(nn.Module):

    def __init__(self, d_in, d_out, context_length,
                 dropout, qkv_bias=False):
        super().__init__()
        self.d_out = d_out
        self.W_query = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.W_key   = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.W_value = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.dropout = nn.Dropout(dropout)
        self.mask = torch.triu(torch.ones(context_length, context_length), diagonal=1)

    def forward(self, x):
        b, num_tokens, d_in = x.shape
        keys = self.W_key(x)
        queries = self.W_query(x)
        values = self.W_value(x)

        attn_scores = queries @ keys.transpose(1, 2)
        attn_scores.masked_fill_(
            self.mask.bool()[:num_tokens, :num_tokens], -torch.inf)
        attn_weights = torch.softmax(
            attn_scores / keys.shape[-1]**0.5, dim=-1
        )
        attn_weights = self.dropout(attn_weights)

        context_vec = attn_weights @ values
        return context_vec
```

可在示例数据上初始化并运行该模块：

```python
torch.manual_seed(123)

inputs = torch.tensor(
  [[0.43, 0.15, 0.89], # Your     (x^1)
   [0.55, 0.87, 0.66], # journey  (x^2)
   [0.57, 0.85, 0.64], # starts   (x^3)
   [0.22, 0.58, 0.33], # with     (x^4)
   [0.77, 0.25, 0.10], # one      (x^5)
   [0.05, 0.80, 0.55]] # step     (x^6)
)

batch = torch.stack((inputs, inputs), dim=0)
context_length = batch.shape[1]
d_in = inputs.shape[1]
d_out = 2

ca_without_buffer = CausalAttentionWithoutBuffers(d_in, d_out, context_length, 0.0)

with torch.no_grad():
    context_vecs = ca_without_buffer(batch)

print(context_vecs)
```

目前一切正常。

但训练 LLM 通常用 GPU 加速。将 `CausalAttentionWithoutBuffers` 模块移到 GPU。

注意：需在配备 GPU 的环境中运行。

```python
has_cuda = torch.cuda.is_available()
has_mps = torch.backends.mps.is_available()

print("Machine has GPU:", has_cuda or has_mps)

if has_mps:
    device = torch.device("mps")   # Apple Silicon GPU (Metal)
elif has_cuda:
    device = torch.device("cuda")  # NVIDIA GPU
else:
    device = torch.device("cpu")   # CPU fallback

print(f"Using device: {device}")

batch = batch.to(device)
ca_without_buffer = ca_without_buffer.to(device)
```

再次运行代码：

```python
with torch.no_grad():
    context_vecs = ca_without_buffer(batch)

print(context_vecs)
```

运行出错。似乎尝试在 GPU 张量与 CPU 张量间做矩阵乘法。但模块已移到 GPU！


再检查一些张量的设备位置：

```python
print("W_query.device:", ca_without_buffer.W_query.weight.device)
print("mask.device:", ca_without_buffer.mask.device)
```

```python
type(ca_without_buffer.mask)
```

可见 `mask` 未移到 GPU，因为它不是 parameter（如 `W_query.weight`）。

需手动 `.to("cuda")` 移到 GPU：

```python
ca_without_buffer.mask = ca_without_buffer.mask.to(device)
print("mask.device:", ca_without_buffer.mask.device)
```

再试一次：

```python
with torch.no_grad():
    context_vecs = ca_without_buffer(batch)

print(context_vecs)
```

这次成功了！

但逐个移张量到 GPU 很繁琐。下一节将看到用 `register_buffer` 注册 `mask` 为 buffer 更简单。

## 使用 buffer 的示例

修改因果注意力类，将因果 `mask` 注册为 buffer：

```python
import torch
import torch.nn as nn

class CausalAttentionWithBuffer(nn.Module):

    def __init__(self, d_in, d_out, context_length,
                 dropout, qkv_bias=False):
        super().__init__()
        self.d_out = d_out
        self.W_query = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.W_key   = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.W_value = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.dropout = nn.Dropout(dropout)
        # Old:
        # self.mask = torch.triu(torch.ones(context_length, context_length), diagonal=1)

        # New:
        self.register_buffer("mask", torch.triu(torch.ones(context_length, context_length), diagonal=1))

    def forward(self, x):
        b, num_tokens, d_in = x.shape
        keys = self.W_key(x)
        queries = self.W_query(x)
        values = self.W_value(x)

        attn_scores = queries @ keys.transpose(1, 2)
        attn_scores.masked_fill_(
            self.mask.bool()[:num_tokens, :num_tokens], -torch.inf)
        attn_weights = torch.softmax(
            attn_scores / keys.shape[-1]**0.5, dim=-1
        )
        attn_weights = self.dropout(attn_weights)

        context_vec = attn_weights @ values
        return context_vec
```

方便的是，模块移到 GPU 时 mask 也会自动在 GPU 上：

```python
ca_with_buffer = CausalAttentionWithBuffer(d_in, d_out, context_length, 0.0)
ca_with_buffer.to(device)

print("W_query.device:", ca_with_buffer.W_query.weight.device)
print("mask.device:", ca_with_buffer.mask.device)
```

```python
with torch.no_grad():
    context_vecs = ca_with_buffer(batch)

print(context_vecs)
```

将张量注册为 buffer 可大大简化：无需手动移张量到 GPU。

## Buffer 与 `state_dict`

- Buffer 相比普通张量的另一优势：包含在模型 `state_dict` 中
- 例如无 buffer 的因果注意力对象的 `state_dict`

```python
ca_without_buffer.state_dict()
```

- 上述 `state_dict` 不含 mask
- 下方因注册为 buffer 而包含 mask

```python
ca_with_buffer.state_dict()
```

- `state_dict` 用于保存/加载训练好的模型
- 本例 mask 训练不变，保存/加载意义不大；为演示假设所有 `1` 改为 `2`：

```python
ca_with_buffer.mask[ca_with_buffer.mask == 1.] = 2.
ca_with_buffer.mask
```

- 保存并加载模型后，mask 会恢复为修改后的值

```python
torch.save(ca_with_buffer.state_dict(), "model.pth")

new_ca_with_buffer = CausalAttentionWithBuffer(d_in, d_out, context_length, 0.0)
new_ca_with_buffer.load_state_dict(torch.load("model.pth"))

new_ca_with_buffer.mask
```

- 若不使用 buffer 则不会如此：

```python
ca_without_buffer.mask[ca_without_buffer.mask == 1.] = 2.

torch.save(ca_without_buffer.state_dict(), "model.pth")

new_ca_without_buffer = CausalAttentionWithoutBuffers(d_in, d_out, context_length, 0.0)
new_ca_without_buffer.load_state_dict(torch.load("model.pth"))

new_ca_without_buffer.mask
```
