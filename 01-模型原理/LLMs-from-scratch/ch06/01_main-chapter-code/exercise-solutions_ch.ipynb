## 练习 6.1：增加上下文长度


可将输入填充至模型支持的最大 token 数，将 `max_length` 设为 1024：

```python
max_length = 1024

train_dataset = SpamDataset(base_path / "train.csv", max_length=max_length, tokenizer=tokenizer)
val_dataset = SpamDataset(base_path / "validation.csv", max_length=max_length, tokenizer=tokenizer)
test_dataset = SpamDataset(base_path / "test.csv", max_length=max_length, tokenizer=tokenizer)
```

或者等价地，可通过以下方式定义 `max_length`：

```python
max_length = model.pos_emb.weight.shape[0]
```

或

```python
max_length = BASE_CONFIG["context_length"]
```


为方便起见，可通过以下命令运行该实验：

```bash
python additional-experiments.py --context_length "model_context_length"
```

## 练习 6.2：微调整个模型


若微调整个模型而非仅最后一个 Transformer 块，可从代码中删除以下行：

```python
for param in model.parameters():
    param.requires_grad = False
```

为方便起见，可通过以下命令运行该实验：

```bash
python additional-experiments.py --trainable_layers all
```

使用 [../02_bonus_additional-experiments](../02_bonus_additional-experiments) 文件夹中的代码；测试准确率提高约 1%，为 96.67%（[ch06_ch.ipynb](ch06_ch.ipynb) 中为 95.67%）。


## 练习 6.3：微调第一个 token 与最后一个 token


若微调第一个输出 token 而非最后一个，可在代码中所有出现的位置将 

```python
model(input_batch)[:, -1, :]
```

改为

```python
model(input_batch)[:, 0, :]
```

为方便起见，可通过以下命令运行该实验：

```
python additional-experiments.py --trainable_token first
```

使用 [../02_bonus_additional-experiments](../02_bonus_additional-experiments) 文件夹中的代码；测试准确率会明显降至 75.00%（[ch06_ch.ipynb](ch06_ch.ipynb) 中为 95.67%）。
